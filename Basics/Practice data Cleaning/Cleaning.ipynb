```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
pd.set_option('future.no_silent_downcasting', True)
```

```python
df = pd.read_csv('Set.csv')
df.info()
```

```python
df['No Violation Found'] = df['No Violation Found'].replace({'No':0,'Yes':1}).infer_objects(copy=False)
df['City Do It'] = df['City Do It'].replace({'No':0,'Yes':1}).infer_objects(copy=False)
df['Owner Will Do It'] = df['Owner Will Do It'].replace({'No':0,'Yes':1}).infer_objects(copy=False)
df['Capital Project Conflict Flag']  = df['Capital Project Conflict Flag'].replace({'No':0,'Yes':1}).infer_objects(copy=False)
df['Cancel'] = df['Cancel'].replace({'No':0,'Yes':1}).infer_objects(copy=False)
df['Inspection Date'] = pd.to_datetime(
    df['Inspection Date'],
    errors='coerce',
    dayfirst=False
)
df['Is 311 Inspection'] = df['Is 311 Inspection'].replace({'No':0,'Yes':1}).infer_objects(copy=False)
df['Pickup Sidewalk'] = df['Pickup Sidewalk'].replace({'No':0,'Yes':1}).infer_objects(copy=False)
df['Curb311'] = df['Curb311'].replace({'No':0,'Yes':1}).infer_objects(copy=False)
df['Pickup curb'] = df['Pickup curb'].replace({'No':0,'Yes':1}).infer_objects(copy=False)
df['Other'] = df['Other'].replace({'No':0,'Yes':1}).infer_objects(copy=False)
df['Correspondence'] = df['Correspondence'].replace({'No':0,'Yes':1}).infer_objects(copy=False)
df['Damage Type Code'] = df['Damage Type Code'].astype('category')
df['Inspection Date'].fillna(pd.Timestamp('1900-01-01'))
```

```python
df.to_csv('new_nigga',index=False)
```
